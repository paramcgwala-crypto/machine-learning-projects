यह एक मशीन लर्निंग प्रोजेक्ट है जिसका मुख्य उद्देश्य मरीज के क्लिनिकल रिकॉर्ड्स (जैसे: Age, Blood Pressure, Cholesterol, Chest Pain Type आदि) का विश्लेषण करके यह प्रेडिक्ट करना है कि मरीज को **Heart Failure (दिल का दौरा)** पड़ने का खतरा है या नहीं। 

**Model Used:** Logistic Regression (Binary Classification)  
**Flow Level:** End-to-End Pipeline (Data Loading ➡️ Exploration ➡️ Dummy Encoding ➡️ Feature Scaling ➡️ Model Training & Evaluation)

Cell 1: Libraries Import Karna

```python
# Data ko table format mein handle karne ke liye pandas import kiya
import pandas as pd

# Data ko training aur testing set mein divide karne ke liye train_test_split liya
from sklearn.model_selection import train_test_split

# Binary classification (0 ya 1) ke liye Logistic Regression model liya
from sklearn.linear_model import LogisticRegression

# Features ko ek saman scale par lane ke liye StandardScaler liya
from sklearn.preprocessing import StandardScaler

# Model ki performance (Accuracy, Precision, Recall, Confusion Matrix) check karne ke liye metrics liye
from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix
```

Cell 2: Data Load Karna

```python
# pd.read_csv se heart.csv file ko read karke df variable mein store kiya
df = pd.read_csv("heart.csv")
```

Cell 3: Shape Check Karna (Rows aur Columns)

```python
# df.shape se check kiya ki pure dataset mein kitne rows aur kitne columns hain

df.shape
```

Cell 4: Data Ki Info Aur Missing Values Check Karna

```python
# df.info() se columns ka data type (int, float, string) aur non-null values ka pata chalta hai

df.info()
```

Cell 5: Head Check Karna (Starting ke 5 Rows)

```python
# df.head(5) se starting ki 5 rows dekh rahe hain taki data ka idea lag sake

df.head(5)
```

Cell 6: Missing Values Ka Sum Dekhna

```python
# Har column mein kitni missing (khaali) values hain, unka total sum check kar rahe hain

print(df.isnull().sum())
```

Cell 7: Target Column Ka Analysis (Kitne log bimar hain aur kitne safe)

```python
# Target column 'HeartDisease' mein 1 (bimar) aur 0 (safe) kitni baar aaya hai, uski ginti dekh rahe hain
print(df["HeartDisease"].value_counts())
```

Cell 8: Gender Ke Hisab Se Heart Disease Rate Dekhna

```python
# Sex ke according data ko group karke average heart disease rate check kar rahe hain

print(df.groupby("Sex")["HeartDisease"].mean())
```

Cell 9: Chest Pain Type Ke Hisab Se Heart Disease Rate Dekhna

```python
# ChestPainType ke according average heart disease rate dekh rahe hain ki kis pain mein kitna risk hai

print(df.groupby("ChestPainType")["HeartDisease"].mean())
```

Cell 10: Preprocessing (Categorical Data Ko Numerical Mein Badalna)

```python
# pd.get_dummies se Text/String columns ko 0 aur 1 format (numerical) mein convert kar rahe hain

df = pd.get_dummies(df, columns=["Sex", "ChestPainType", "RestingECG", "ExerciseAngina", "ST_Slope"], drop_first=True)
```

Cell 11: Preprocessing Ke Baad Columns Dekhna

```python
# get_dummies apply karne ke baad columns ke naye naam check kar rahe hain

df.columns
```

Cell 12: Features (X) Aur Target (y) Algu Karna Aur Split Karna

```python
# X mein se target column 'HeartDisease' ko drop karke saare input features rakh rahe hain
X = df.drop("HeartDisease", axis=1)

# y mein sirf target column 'HeartDisease' rakh rahe hain jisko predict karna hai
y = df["HeartDisease"]

# Data ko 80% training aur 20% testing mein divide kar rahe hain, random_state se split hamesha same rahega
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# Training aur testing data ka final size check kar rahe hain
print("Training:", X_train.shape)
print("Testing:", X_test.shape)
```

Cell 13: Feature Scaling (StandardScaler)

```python
# Scaler ka object banaya taki bade numerical values (jaise Age, Bp, Cholesterol) ko same scale par la sakein
scaler = StandardScaler()

# X_train par fit karke mean aur variance calculate kiya aur fir data ko scale kar diya
X_train = scaler.fit_transform(X_train)

# X_test ko bina fit kiye, train wale same scale par transform (scale) kar rahe hain
X_test = scaler.transform(X_test)
```

Cell 14: Model Train Karna Aur Predictions Nikalna

```python
# Logistic Regression model ka object banaya aur max_iter=1000 diya taki model aache se converge ho sake
model = LogisticRegression(max_iter=1000)

# model.fit se scale kiye huye training data par model ko train kar rahe hain
model.fit(X_train, y_train)

# Train data par model se prediction karwa rahe hain
y_train_pred = model.predict(X_train)

# Test data par model se final prediction karwa rahe hain
y_test_pred = model.predict(X_test)
```

Cell 15: Evaluation Metrics (Accuracy, Precision, Recall)

```python
# Training data ki accuracy, precision aur recall score calculate kar rahe hain
train_accuracy = accuracy_score(y_train, y_train_pred)
train_precision = precision_score(y_train, y_train_pred)
train_recall = recall_score(y_train, y_train_pred)

# Testing data ki accuracy, precision aur recall score calculate kar rahe hain
test_accuracy = accuracy_score(y_test, y_test_pred)
test_precision = precision_score(y_test, y_test_pred)
test_recall = recall_score(y_test, y_test_pred)

# Saare calculated scores ko terminal par print kar rahe hain
print("Training Accuracy:", train_accuracy)
print("Training Precision:", train_precision)
print("Training Recall:", train_recall)
print("Testing Accuracy:", test_accuracy)
print("Testing Precision:", test_precision)
print("Testing Recall:", test_recall)
```

Cell 16: Confusion Matrix

```python
# Training data ka confusion matrix nikaal kar True Negative, False Positive, False Negative aur True Positive alag kiya
train_cm = confusion_matrix(y_train, y_train_pred)
TN_train, FP_train, FN_train, TP_train = train_cm.ravel()

# Training confusion matrix print kar rahe hain
print("Training Confusion Matrix:")
print(train_cm)
print("TN:", TN_train, "FP:", FP_train, "FN:", FN_train, "TP:", TP_train)

# Testing data ka confusion matrix nikaal kar test values alag kiye
test_cm = confusion_matrix(y_test, y_test_pred)
TN_test, FP_test, FN_test, TP_test = test_cm.ravel()

# Testing confusion matrix print kar rahe hain
print("\nTesting Confusion Matrix:")
print(test_cm)
print("TN:", TN_test, "FP:", FP_test, "FN:", FN_test, "TP:", TP_test)
```
